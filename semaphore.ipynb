# Semaphore API integration tests

This notebook runs Semaphore's public/user-visible APIs, such as listing broadcasts, to verify that the service is running.

```python
import os
from httpx import AsyncClient
from lsst.rsp import RSPInternalDiscovery
```

## Find Semaphore URL

```python
discovery = RSPInternalDiscovery()
semaphore_url = discovery.get_internal_service_url("semaphore")
print(f"Semaphore base URL: {semaphore_url}")
```

## Create authenticated client

```python
session = discovery.get_session()
```

## Access Semaphore app metadata

```python
r = session.get(semaphore_url + "/")
metadata = r.json()
print(f"Semaphore version {metadata['metadata']['version']}")
```

## Check broadcasts endpoint

```python
r = session.get(semaphore_url + "/v1/broadcasts")
broadcasts = r.json()
assert isinstance(broadcasts, list)
```
