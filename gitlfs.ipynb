# Git LFS test

This notebook checks out a git-lfs-backed notebook and verifies that it retrieved the (LFS-stored) contents, rather than the Git LFS header.

## Imports and constants

```python
import os
import shutil
import subprocess
from tempfile import TemporaryDirectory
from pathlib import Path

repo="tutorial-notebooks-data"
gh_url=f"https://github.com/lsst/{repo}"
testfile = "PGC_038749_I_g_bbl2011.fits"
```

## Save some debugging output

```python
home=os.environ["HOME"]
path=os.environ["PATH"]
git=shutil.which("git")
gitlfs=shutil.which("git-lfs")

dbg_txt=f"HOME={home} ; git={git} ; git-lfs={gitlfs}\nPATH={path}"
print(dbg_txt)
```

## Clone repo and read first six bytes of test file

```python
with TemporaryDirectory() as td:
    repo_path = Path(td) / repo
    cmd=[ "git", "clone", gh_url, str(repo_path) ]  
    proc=subprocess.run(cmd, capture_output=True)

    tf = repo_path / "data" / testfile
    
    with tf.open("rb") as f:
        sixbytes = f.read(6)
```

## Verify we have contents, not stub file

```python
sixstr = sixbytes.decode()
output = f"{dbg_txt}\nExpected 'SIMPLE', received '{sixstr}'; stdout={proc.stdout.decode()},stderr={proc.stderr.decode()}"
assert sixstr == "SIMPLE", output
print(output)  # Just for debugging
```
